# Data Extraction and Initial Probing
This notebook handles the initial ingestion of the raw CSV files and performs basic checks on their structure and contents.


```python
%pip install pandas numpy

import pandas as pd
import numpy as np
import os
```

## 1. Load Raw Datasets


```python
urls = {
    'calendar': 'https://drive.google.com/uc?export=download&id=1jnbXUjOaWxb6oTQh7ZGc8hrsvRbBTZzz',
    'customers': 'https://drive.google.com/uc?export=download&id=16iirMkTGEWdqge_tNs18hkBY-hBrigcD',
    'products': 'https://drive.google.com/uc?export=download&id=1hFeWT03Cr6EImUzO-bTJ-i2d7q3sp83Z',
    'sales': 'https://drive.google.com/uc?export=download&id=1vrypPferlfcZFJTTBKExYAvLxvat6w0Z',
    'stores': 'https://drive.google.com/uc?export=download&id=1cGYO7XGxZd11lz7m8PxiRKssYgkwJGtW'
}

dataframes = {}
for table_name, url in urls.items():
    dataframes[table_name] = pd.read_csv(url)
    print(f"Loaded {table_name} from Google Drive.")
```

## 2. Basic Information (Info & Shape)


```python
for name, df in dataframes.items():
    print(f"\n{'='*50}")
    print(f"TABLE: {name.upper()} (Shape: {df.shape})")
    print(f"{'='*50}")
    df.info()
    print("\n")
```

## 3. Data Preview (Head)


```python
for name, df in dataframes.items():
    print(f"\n--- First 3 rows of {name.upper()} ---")
    display(df.head(3))
```

## 4. Descriptive Statistics


```python
for name, df in dataframes.items():
    print(f"\n--- Summary Stats for {name.upper()} ---")
    display(df.describe(include='all').T)
```

## 5. Missing Values Check


```python
for name, df in dataframes.items():
    missing = df.isnull().sum()
    if missing.sum() > 0:
        print(f"\n--- Missing Values in {name.upper()} ---")
        print(missing[missing > 0])
    else:
        print(f"{name.upper()}: No missing values detected.")
```
